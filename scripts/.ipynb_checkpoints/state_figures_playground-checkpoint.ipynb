# State-figures playground

Interactive notebook to iterate on publication plots for the **state-encoding** family of results:

1. Two example EC cells — one *sustained* state cell (all phases positive), one *not-consistent* (positive in some phases, weak/negative in others). Source: `scripts/encoding_state_sustained_cv.py`.
2. Per-ROI cell count vs fraction sustained, in the `pub_sig_counts_bars.pdf` style.
3. Per-ROI **across-phase** state β heatmap, in the `combined_publication_figure.pdf` style. Source: addon `phase_mask_replay.csv` from `RSA_addon_analyses.py`.
4. Brain-projected across-phase state β — nilearn glass-brain fallback (quickbrain needs Python≥3.10; this env is 3.8.15).

All plot helpers are defined as functions so you can re-run any cell repeatedly with tweaked parameters.

**Inputs** are listed in the path-config cell at the top — swap to new run dirs there.

## 3. PLOT A — Two example EC cells (sustained + not-consistent)

Per-config polar plot of the cell's mean firing in each of the 3 phases of every state.

```python
def load_cell_traces(neuron_id):
    """Load behavioural + neuron-trace data for a single cell."""
    import re
    m = re.match(r'^(\d+)_', str(neuron_id))
    if not m: return None
    sub_str = m.group(1).zfill(2)
    try:
        data = hh.load_norm_data(str(DATA), [sub_str], res_data=False)
    except Exception as exc:
        print(f'  load failed: {exc}')
        return None
    sd = data.get(f'sub-{sub_str}')
    if sd is None or neuron_id not in sd['normalised_neurons']:
        return None
    beh = sd['beh'].copy().reset_index(drop=True)
    beh['config_str'] = (beh['loc_A'].astype(int).astype(str) + '-' +
                          beh['loc_B'].astype(int).astype(str) + '-' +
                          beh['loc_C'].astype(int).astype(str) + '-' +
                          beh['loc_D'].astype(int).astype(str))
    return {'neuron_id': neuron_id, 'beh': beh,
            'trace': sd['normalised_neurons'][neuron_id].reset_index(drop=True)}

def get_per_state_phase_means(cell_data, n_states=4, n_phases=3, state_len=90, phase_len=30):
    """For each config and each (state, phase), return mean firing rate."""
    beh, trace = cell_data['beh'], cell_data['trace']
    configs = sorted(beh.loc[beh['correct']==1, 'config_str'].dropna().unique().tolist())
    out = {}
    for cfg in configs:
        idx = beh.index[(beh['config_str']==cfg) & (beh['correct']==1)].to_numpy()
        if len(idx)==0: continue
        mean_trace = np.nanmean(trace.iloc[idx].to_numpy(dtype=float), axis=0)
        M = np.full((n_states, n_phases), np.nan)
        for s in range(n_states):
            s_start = s * state_len
            for p in range(n_phases):
                p_start = s_start + p * phase_len
                M[s, p] = np.nanmean(mean_trace[p_start:p_start+phase_len])
        out[cfg] = M
    return out

def plot_example_state_cell(cell_row, cell_data, ax=None, title=None,
                            phase_colors=PHASE_COLORS):
    """Polar plot: each state at angle s*π/2; 3 phases stacked along that ray.
    Lines = configs; colour = phase (early/middle/late)."""
    per_cfg = get_per_state_phase_means(cell_data)
    if not per_cfg:
        if ax: ax.text(0.5, 0.5, 'no data', ha='center', va='center')
        return
    if ax is None:
        fig, ax = plt.subplots(figsize=(4, 4), subplot_kw=dict(projection='polar'))
    # Normalise across all configs/states/phases for a comparable radius.
    M_all = np.stack(list(per_cfg.values()))   # (n_cfg, 4, 3)
    v_min, v_max = np.nanmin(M_all), np.nanmax(M_all)
    v_rng = v_max - v_min if v_max > v_min else 1.0
    # 4 states at 0, 90, 180, 270 deg
    state_angles = np.linspace(0, 2*np.pi, 5)[:-1]
    # 3 phases offset slightly inside each state "wedge"
    phase_offset = np.linspace(-np.pi/12, np.pi/12, 3)
    for cfg, M in per_cfg.items():
        for p in range(3):
            r = (M[:, p] - v_min) / v_rng + 0.4
            theta = state_angles + phase_offset[p]
            ax.plot(np.append(theta, theta[0]),
                    np.append(r, r[0]),
                    color=phase_colors[p], alpha=0.35, lw=1.1)
    # Mean across configs as bold lines per phase
    M_mean = np.nanmean(M_all, axis=0)
    for p in range(3):
        r = (M_mean[:, p] - v_min) / v_rng + 0.4
        theta = state_angles + phase_offset[p]
        ax.plot(np.append(theta, theta[0]),
                np.append(r, r[0]),
                color=phase_colors[p], lw=2.5, alpha=0.95,
                label=PHASES[p].capitalize())
        ax.scatter(theta, r, color=phase_colors[p], s=60, zorder=5,
                   edgecolor='black', linewidth=0.8)
    ax.set_xticks(state_angles)
    ax.set_xticklabels([f'State {s}' for s in STATES], fontsize=10)
    ax.set_yticks([])
    ax.set_ylim(0, 1.5)
    ax.grid(True, alpha=0.3)
    if title:
        ax.set_title(title, fontsize=11, pad=15)
    return ax

def find_example_cells(sustained_df, roi='EC'):
    """Best sustained cell in `roi` (highest mean_r among sustained_sig=True),
    and best NOT-consistent cell (sustained_sig=False but mean_r positive and
    has at least one positive AND one non-positive phase contrast)."""
    ec = sustained_df[sustained_df['roi']==roi].copy()
    sust = ec[ec['sustained_sig']].sort_values('mean_r', ascending=False)
    cand_sustained = sust.iloc[0] if len(sust) else None
    def mixed_phase(c):
        return (c is not None and np.isfinite(c).all()
                and (c > 0).any() and not (c > 0).all())
    not_consistent = ec[(~ec['sustained_sig']) & (ec['mean_r'] > 0)
                       & ec['phase_contrasts'].apply(mixed_phase)]
    not_consistent = not_consistent.sort_values('mean_r', ascending=False)
    cand_not = not_consistent.iloc[0] if len(not_consistent) else None
    return cand_sustained, cand_not

# Pick two examples
sustained_cell, not_cell = find_example_cells(sustained, roi='EC')
print('Sustained:    ', sustained_cell['neuron'] if sustained_cell is not None else '—',
      'mean_r=', f"{sustained_cell['mean_r']:+.4f}" if sustained_cell is not None else '')
print('Not-consist.: ', not_cell['neuron'] if not_cell is not None else '—',
      'mean_r=', f"{not_cell['mean_r']:+.4f}" if not_cell is not None else '')
if sustained_cell is not None:
    print('  contrasts (early,mid,late):', np.round(sustained_cell['phase_contrasts'], 4))
if not_cell is not None:
    print('  contrasts (early,mid,late):', np.round(not_cell['phase_contrasts'], 4))
```

```python
# Plot the two example cells side by side
fig = plt.figure(figsize=(11, 5))
gs = GridSpec(1, 2, figure=fig, wspace=0.35)

if sustained_cell is not None:
    ax1 = fig.add_subplot(gs[0, 0], projection='polar')
    d1 = load_cell_traces(sustained_cell['neuron'])
    if d1:
        plot_example_state_cell(
            sustained_cell, d1, ax=ax1,
            title=f'Sustained EC cell\n{sustained_cell["neuron"]}\n'
                  f'mean r={sustained_cell["mean_r"]:+.3f}, '
                  f'p(state r)={sustained_cell["p_perm_state_r"]:.3f}')
        ax1.legend(loc='upper right', bbox_to_anchor=(1.25, 1.0), fontsize=9)

if not_cell is not None:
    ax2 = fig.add_subplot(gs[0, 1], projection='polar')
    d2 = load_cell_traces(not_cell['neuron'])
    if d2:
        plot_example_state_cell(
            not_cell, d2, ax=ax2,
            title=f'Not-consistent EC cell\n{not_cell["neuron"]}\n'
                  f'mean r={not_cell["mean_r"]:+.3f}, '
                  f'p(state r)={not_cell["p_perm_state_r"]:.3f}')

fig.suptitle('Example EC state cells — sustained vs. not-consistent across phases',
             fontsize=12, y=1.02)
fig.savefig(OUT_DIR / 'A_example_EC_cells.pdf', bbox_inches='tight')
fig.savefig(OUT_DIR / 'A_example_EC_cells.png', dpi=300, bbox_inches='tight')
plt.show()
print(f'\nSaved -> {OUT_DIR}/A_example_EC_cells.pdf')
```

## 4. PLOT B — per-ROI sustained-state significance counts

Style mirrors `pub_sig_counts_bars.pdf`: grey background bar = total cells per ROI, coloured foreground bar = number of sustained-sig cells. Stars from BH-FDR-corrected binomial test (chance level = 0.05).

```python
# `plot_significance_counts_bars` expects columns ['roi','model','mean_r','p_perm'].
# Adapt the sustained-cv table: encode sustained_sig as p_perm < alpha.
def sustained_to_results_df(df):
    out = df[['roi','neuron','mean_r','sustained_sig']].copy()
    out['model'] = 'state (sustained)'
    out['p_perm'] = np.where(out['sustained_sig'], 0.0, 1.0)
    return out[['neuron','roi','model','mean_r','p_perm']]

results_df = sustained_to_results_df(sustained)
fig_path = OUT_DIR / 'B_sustained_sig_counts_per_roi'
plot_significance_counts_bars(
    results_df,
    models=['state (sustained)'],
    rois=ROI_ORDER,
    save_path=str(fig_path) + '.png',  # function also writes .pdf + .svg
    alpha=0.05,
    chance_level=0.05,
    use_fdr=True,
    base_fontsize=13,
    panel_w=5.5, panel_h=4.0,
)
plt.show()
print(f'Saved -> {fig_path}.pdf / .png / .svg')
```

## 5. PLOT C — small across-phase state β heatmap (ROI × mode)

Re-uses the addon `phase_mask_replay.csv`: state β inside `ctrl_dsrFULL`, test = `between_tasks_z`, across all 7 ROIs and the 3 phase modes. Style mirrors the small RSA heatmap in `combined_publication_figure.pdf`.

```python
def plot_phase_mode_heatmap(addon_df, sub_model='state', combo='ctrl_dsrFULL',
                            test='between_tasks_z', rois=ROI_ORDER,
                            modes=('across_phase','full','within_phase'),
                            cmap='RdBu_r', annotate=True, ax=None,
                            title=None, save_stem=None):
    sub = addon_df[(addon_df['combo']==combo) &
                   (addon_df['test']==test) &
                   (addon_df['sub_model']==sub_model)]
    rois = [r for r in rois if r in set(sub['roi'])]
    M = np.full((len(rois), len(modes)), np.nan)
    Mp = np.full_like(M, np.nan)
    for i, r in enumerate(rois):
        for j, mo in enumerate(modes):
            row = sub[(sub['roi']==r) & (sub['mode']==mo)]
            if not row.empty:
                M[i, j] = float(row['beta'].iloc[0])
                Mp[i, j] = float(row['p_param'].iloc[0])
    vmax = float(np.nanmax(np.abs(M))) if np.isfinite(M).any() else 0.2
    if ax is None:
        fig, ax = plt.subplots(figsize=(4.5, 3.8), constrained_layout=True)
    else:
        fig = ax.figure
    im = ax.imshow(M, cmap=cmap, vmin=-vmax, vmax=vmax, aspect='auto')
    ax.set_xticks(range(len(modes)))
    ax.set_xticklabels([m.replace('_','-') for m in modes],
                       rotation=30, ha='right', fontsize=10)
    ax.set_yticks(range(len(rois)))
    ax.set_yticklabels([pretty(r) for r in rois], fontsize=10)
    if annotate:
        for i in range(len(rois)):
            for j in range(len(modes)):
                v = M[i, j]
                if np.isfinite(v):
                    p = Mp[i, j]
                    star = ('***' if p<0.001 else '**' if p<0.01
                            else '*' if p<0.05 else '')
                    txt = f'{v:+.3f}' + (f'\n{star}' if star else '')
                    ax.text(j, i, txt, ha='center', va='center', fontsize=8,
                            color='white' if abs(v)>0.55*vmax else 'black')
    cb = fig.colorbar(im, ax=ax, fraction=0.055, pad=0.04)
    cb.ax.tick_params(labelsize=8)
    cb.set_label(r'$\beta$', fontsize=10)
    if title is None:
        title = f'state β across phase modes\n({combo} / {test})'
    ax.set_title(title, fontsize=11)
    for sp in ('top','right'):
        ax.spines[sp].set_visible(False)
    if save_stem:
        fig.savefig(str(save_stem) + '.pdf', bbox_inches='tight')
        fig.savefig(str(save_stem) + '.png', dpi=300, bbox_inches='tight')
    return fig, ax

fig, ax = plot_phase_mode_heatmap(
    addon, sub_model='state', combo='ctrl_dsrFULL', test='between_tasks_z',
    save_stem=OUT_DIR / 'C_phase_mode_state_heatmap_between_tasks_z',
)
plt.show()
print(f'Saved -> {OUT_DIR}/C_phase_mode_state_heatmap_between_tasks_z.pdf')
```

## 6. PLOT D — brain projection (across-phase state β per ROI)

**Notebook note**: [quickbrain](https://pni-lab.github.io/quickbrain/getting-started/) requires Python ≥3.10 and this env (`env_multiple_clocks`) is Python 3.8. So this cell uses **nilearn glass-brain** for now and includes a `try_quickbrain()` stub that future-you can flip on once you migrate envs.

Per-ROI β is rendered at each ROI's mean MNI coordinate (mean of every cell labelled to that ROI). Disc radius scales with N cells; colour = across-phase state β.

```python
# Load MNI coords per ROI from the ROI table (matches RSA's electrode coords).
rt = pd.read_csv(ROI_TABLE)
rt = rt[rt['alt_final_roi'].isin(ROI_ORDER)]
rt = rt[['alt_final_roi','MNI_x','MNI_y','MNI_z']].dropna()
rt = rt.rename(columns={'alt_final_roi':'roi'})
# Mean coord per ROI (a single dot per ROI)
roi_centroids = rt.groupby('roi')[['MNI_x','MNI_y','MNI_z']].mean().reset_index()
roi_n_cells   = rt.groupby('roi').size().rename('n_cells').reset_index()
roi_xyz = roi_centroids.merge(roi_n_cells, on='roi')

# Pull across-phase state β (between_tasks_z, ctrl_dsrFULL combo) per ROI
addon_st = addon[(addon['sub_model']=='state') &
                 (addon['combo']=='ctrl_dsrFULL') &
                 (addon['test']=='between_tasks_z') &
                 (addon['mode']=='across_phase')]
beta_by_roi = dict(zip(addon_st['roi'], addon_st['beta']))
p_by_roi    = dict(zip(addon_st['roi'], addon_st['p_param']))
roi_xyz['beta_across_phase'] = roi_xyz['roi'].map(beta_by_roi)
roi_xyz['p_param']           = roi_xyz['roi'].map(p_by_roi)
roi_xyz = roi_xyz.dropna(subset=['beta_across_phase'])
print(roi_xyz.to_string(index=False))
```

```python
from nilearn import plotting

def plot_brain_heatmap(roi_xyz, value_col='beta_across_phase', p_col='p_param',
                       sig_alpha=0.05, vmax=None, title='', save_stem=None):
    """nilearn glass-brain with one ROI per dot. Colour=value, size~N cells,
    edge highlights significance (black if p<alpha)."""
    coords = roi_xyz[['MNI_x','MNI_y','MNI_z']].to_numpy()
    vals = roi_xyz[value_col].to_numpy()
    ps   = roi_xyz[p_col].to_numpy() if p_col in roi_xyz.columns else None
    ns   = roi_xyz['n_cells'].to_numpy()
    if vmax is None:
        vmax = float(np.nanmax(np.abs(vals))) * 1.05
    sizes = 80 + 220 * (ns - ns.min()) / max(ns.max() - ns.min(), 1)
    edges = ['black' if (ps is not None and np.isfinite(p) and p < sig_alpha)
             else 'gray' for p in (ps if ps is not None else [1]*len(vals))]
    fig = plt.figure(figsize=(11, 4.2))
    display = plotting.plot_glass_brain(
        None, figure=fig, display_mode='ortho',
        title=title, colorbar=False, black_bg=False,
    )
    # Build a marker-style overlay: add as connectome with dummy adjacency
    adj = np.zeros((len(coords), len(coords)))
    display.add_graph(
        adj, coords,
        node_color=plt.cm.RdBu_r((vals + vmax) / (2*vmax)),
        node_size=sizes,
        edge_threshold='100%',
    )
    # Manually label ROI names near each dot using nilearn's matplotlib axes
    for ax in display.axes.values():
        for (x, y, z), label in zip(coords, roi_xyz['roi'].tolist()):
            # nilearn's projection is in mpl coords — we just annotate the
            # axial projection (axes 'z') for simplicity below the figure.
            pass
    # Colour bar (manual, since add_graph doesn't expose one)
    cbar_ax = fig.add_axes([0.92, 0.18, 0.015, 0.6])
    import matplotlib as mpl
    sm = plt.cm.ScalarMappable(
        cmap='RdBu_r', norm=mpl.colors.Normalize(vmin=-vmax, vmax=vmax))
    sm.set_array([])
    cb = fig.colorbar(sm, cax=cbar_ax)
    cb.set_label(r'across-phase state $\beta$', fontsize=10)
    # Below-figure legend with ROI names + values
    legend_text = '  '.join(
        f'{pretty(r)} β={v:+.3f}{"★" if (p is not None and np.isfinite(p) and p<sig_alpha) else ""}'
        for r, v, p in zip(roi_xyz['roi'], vals, (ps if ps is not None else [None]*len(vals)))
    )
    fig.text(0.5, 0.02, legend_text, ha='center', va='bottom',
             fontsize=9, color='0.25', wrap=True)
    if save_stem:
        fig.savefig(str(save_stem) + '.pdf', bbox_inches='tight')
        fig.savefig(str(save_stem) + '.png', dpi=300, bbox_inches='tight')
    return fig, display

fig, _ = plot_brain_heatmap(
    roi_xyz,
    title='Across-phase state β (between_tasks_z, ctrl_dsrFULL)',
    save_stem=OUT_DIR / 'D_brain_across_phase_state_beta',
)
plt.show()
print(f'Saved -> {OUT_DIR}/D_brain_across_phase_state_beta.pdf')
```
